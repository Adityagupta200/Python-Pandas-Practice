```python
import pandas as pd

df = pd.DataFrame({
    "Age" : [22, 35, 29, 48, None],
    "Fare" : [7.25, 71.28, 7.93, 53.10, 8.05]
})
print(df["Age"].mean())
print(df.median())
print(df[["Age", "Fare"]].describe())
```

# Grouped Stats

```python
titanic = pd.DataFrame({
    "sex" : ["female", "male", "male", "female"],
    "age" : [27.9, 30.7, 22.0, 35.0],
    "fare" : [44.5, 25.5, 7.9, 106.1]
})

print(titanic.groupby("sex")["age"].mean())
print(titanic.groupby(["sex"]).mean(numeric_only=True))
```

# Frequency counts

```python
s = pd.Series(["A", "B", "A", "C", "B", "B", None])
print(s.value_counts()) # Count excluding NaNs
print(s.value_counts(dropna=False)) # Count including NaNs
```

# Custom Aggregations

```python
df = pd.DataFrame({
    "Age": [22, 35, 29, 40],
    "Fare": [7.25, 71.28, 7.93, 53.10]
})

df_agg = df.agg({
    "Age" : ["mean", "min", "max", "skew", "count"],
    "Fare" : ["mean", "min", "max", "skew", "count"]
})

df_agg
```

# Computing changes over time with .pct_change()

```python
import numpy as np
s = pd.Series([1,2,3,4,5,4])

print(s.pct_change())
```

```python
df = pd.DataFrame(np.random.randn(5,2))

print(df.pct_change()) # Row-wise changes
print(df.pct_change(axis=1)) # Column-wise changes
```

# Correlation and covariance

```python
frame = pd.DataFrame(np.random.randn(10, 3), columns= ["a", "b", "c"])
print(frame)
cov_a_b = frame["a"].cov(frame["b"])
print(f"\nCovariance between column 'a' and 'b' is {cov_a_b}")
print(frame.cov())
```

# Ranking

```python
s = pd.Series([10, 30, 20, 30])
print(s.rank())                # Average rank on ties
print(s.rank(method="min"))    # Lowest rank on ties
print(s.rank(ascending=False)) # Reverse ranking
```

# Rolling windows

```python
s = pd.Series(range(5))
print(s.rolling(window=2).sum()) # fixed-size rolling sum

s2 = pd.Series(range(5), index= pd.date_range('2025-01-01', periods=5, freq='D'))
s2.rolling(window="2D").sum()
```

# Expanding windows

```python
df = pd.DataFrame({"x": [0,1,2,3,4]})
print(df.expanding(min_periods=2).mean()) # cumlative mean
```

# Exponentially weighted

```python
df = pd.DataFrame({"x" : [0, 1, 2, np.nan, 4]})
times = pd.date_range("2025/01/01", periods=5, freq="D")
df.ewm(halflife= "4 days", times=times).mean()
```
