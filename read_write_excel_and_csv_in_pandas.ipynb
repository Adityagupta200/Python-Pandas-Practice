# Reading csv files

```python
import pandas as pd
df = pd.read_csv("data/stock_data.csv", skiprows=1) # using skipsrows argument to skip the redundant header
df
```

```python
df = pd.read_csv("data/stock_data.csv", header=None, names=["tickers","eps","revenue","price","people"]) # Making a header and specifying column names
df
```

```python
df = pd.read_csv("data/stock_data.csv", nrows=4) # Reading 4 rows
df
```

```python
df = pd.read_csv("data/stock_data.csv", na_values=["not available", "n.a."]) # replacing ["not available", "n.a."] with NaN values
df
```

```python
# Replacing ["not available", "n.a."] and other irrelevant values with NaN
df = pd.read_csv("data/stock_data.csv", na_values = {
    'eps' : ["not available", "n.a."],
    'revenue' : ["not available", "n.a.", "-1"],
    'people' : ["not available", "n.a."],
    'price' : ["not available", "n.a.", "-1"]
})
df
```

# Writing to CSV

```python
df.to_csv("data/new.csv", index=False)
```

```python
df.to_csv("data/new.csv", columns = ["tickers", "eps"], index=False) # Exporting only the specified colums to new.csv
```

# Reading Excel files

```python
df = pd.read_excel("data/new.xlsx")
df
```

```python
# Using "converter=" argument
def convert_people_cell(cell):
    if cell == "n.a.":
        return "Sam Walton"
    return cell

def convert_eps_cell(cell):
    if cell == "not available":
        return None
    return cell
df = pd.read_excel("data/new.xlsx", converters={
    "people" : convert_people_cell,
    "eps" : convert_eps_cell
}, index_col=0)
df
```

```python
df.to_excel("data/new_converted.xlsx", index=False)
df_converted = pd.read_excel("data/new_converted.xlsx", index_col=0)
df_converted
```

```python
df.to_excel("data/new_converted.xlsx", index= False, startrow=1, startcol=2, sheet_name="stocks")
df_converted = pd.read_excel("data/new_converted.xlsx", index_col=0)
df_converted
```

```python
#
df_stocks = pd.read_excel("data/new_converted.xlsx", index_col=0)
df_weather = pd.read_csv("data/weather_by_cities.csv", index_col=0)

with pd.ExcelWriter("data/stocks_weather.xlsx") as writer:
    df_stocks.to_excel(writer, sheet_name="stocks")
    df_weather.to_excel(writer, sheet_name="weather")
```
